```python
import re
import json
from typing import Dict, List, Tuple
from dataclasses import dataclass

@dataclass
class QualityMetrics:
    metrics: Dict[str, float]

    def overall_score(self) -> float:
        if not self.metrics:
            return 0.0
        return sum(self.metrics.values()) / len(self.metrics)

class SelfReflectiveImprover:
    def __init__(self):
        self.improvement_history = []
        self.reflection_criteria = {
            'clarity': 'Is the content clear and easy to understand?',
            'completeness': 'Does it cover all important aspects?',
            'accuracy': 'Is the information factually correct?',
            'relevance': 'Is it relevant to the user\'s needs?',
            'engagement': 'Is it engaging and well-structured?'
        }

    def improve_with_reflection(self, initial_content: str, content_type: str = "summary") -> Dict:
        """Main reflection and improvement loop"""

        current_content = initial_content
        iteration = 1
        max_iterations = 3

        print(f"=== INITIAL CONTENT ===\\n{current_content}\\n")

        while iteration <= max_iterations:
            print(f"=== REFLECTION ITERATION {iteration} ===")

            # Step 1: Self-critique
            critique = self._generate_self_critique(current_content, content_type)
            print(f"🔍 SELF-CRITIQUE:\\n{critique}\\n")

            # Step 2: Quality assessment
            quality_scores = self._assess_quality(current_content, critique)
            print(f"📊 QUALITY SCORES:\\n{self._format_quality_scores(quality_scores)}\\n")

            # Step 3: Check if improvement is needed
            if quality_scores.overall_score() >= 8.0:
                print("✅ Quality threshold reached. No further improvement needed.\\n")
                break

            # Step 4: Generate improvements
            improved_content = self._generate_improvements(current_content, critique, quality_scores)
            print(f"🔄 IMPROVED CONTENT:\\n{improved_content}\\n")

            # Step 5: Compare with previous version
            comparison = self._compare_versions(current_content, improved_content)
            print(f"📈 IMPROVEMENT ANALYSIS:\\n{comparison}\\n")

            # Update for next iteration
            current_content = improved_content
            iteration += 1

        return {
            'final_content': current_content,
            'improvement_history': self.improvement_history,
            'final_quality_score': quality_scores.overall_score()
        }

    def _generate_self_critique(self, content: str, content_type: str) -> str:
        """Generate detailed self-critique"""
        critique_prompt = f"""
        Content Type: {content_type}
        Content to Critique: "{content}"

        Provide a detailed critique of this content based on:
        1. CLARITY: Is it easy to understand? Any confusing parts?
        2. COMPLETENESS: What important information is missing?
        3. ACCURACY: Any factual errors or misleading information?
        4. RELEVANCE: Does it address the main topic effectively?
        5. ENGAGEMENT: Is it interesting and well-structured?

        Be specific about what needs improvement and why.
        Rate each aspect from 1-10 and provide concrete suggestions.
        """

        # Simulate self-critique
        critique = f"""
        SELF-CRITIQUE ANALYSIS:

        1. CLARITY (Score: 7/10):
           - Generally clear but some technical terms could be explained better
           - Sentence structure is mostly good but a few long sentences
           - Could benefit from more examples or analogies

        2. COMPLETENESS (Score: 6/10):
           - Missing some key details that users would find helpful
           - Could include more practical applications or examples
           - Some logical gaps between concepts

        3. ACCURACY (Score: 8/10):
           - Information appears factually correct
           - Could use more specific data or citations
           - Some generalizations that could be more precise

        4. RELEVANCE (Score: 7/10):
           - Addresses the main topic but could be more focused
           - Some tangential information that could be condensed
           - Core message is relevant but could be clearer

        5. ENGAGEMENT (Score: 6/10):
           - Somewhat dry presentation
           - Could use better formatting and structure
           - Needs more compelling introduction and conclusion

        OVERALL ASSESSMENT: Content has solid foundation but needs refinement
        for clarity, completeness, and engagement.
        """

        return critique

    def _assess_quality(self, content: str, critique: str) -> QualityMetrics:
        """Extract quality scores from critique"""
        # Parse scores from critique text
        scores = {}
        for criterion in self.reflection_criteria.keys():
            # Look for score patterns in critique
            pattern = rf"{criterion.upper()}.*?(\d+)/10"
            match = re.search(pattern, critique, re.IGNORECASE)
            if match:
                scores[criterion] = float(match.group(1))
            else:
                scores[criterion] = 5.0  # Default score

        return QualityMetrics(metrics=scores)

    def _generate_improvements(self, content: str, critique: str, quality_scores: QualityMetrics) -> str:
        """Generate improved version based on critique"""
        improvement_prompt = f"""
        Original Content: {content}

        Critique Analysis: {critique}

        Quality Scores: {self._format_quality_scores(quality_scores)}

        Based on the critique, rewrite the content to address the specific issues mentioned:

        1. Improve CLARITY by simplifying language and adding examples
        2. Enhance COMPLETENESS by adding missing information
        3. Ensure ACCURACY with precise statements
        4. Increase RELEVANCE by staying focused on key points
        5. Boost ENGAGEMENT with better structure and flow

        Provide a significantly improved version that addresses each critique point.
        """

        # Simulate improved content generation
        improved_content = f"""
        IMPROVED VERSION:

        {content}

        [Enhanced with:]
        - Clearer explanations and examples
        - Additional relevant details and context
        - Better structure and formatting
        - More engaging introduction and conclusion
        - Improved flow between concepts
        - Specific actionable insights
        """

        # Store improvement record
        improvement_record = {
            'iteration': len(self.improvement_history) + 1,
            'original_scores': quality_scores,
            'improvements_made': [
                "Enhanced clarity with examples",
                "Added missing contextual information",
                "Improved structure and flow",
                "Made content more engaging"
            ]
        }

        self.improvement_history.append(improvement_record)

        return improved_content

    def _compare_versions(self, original: str, improved: str) -> str:
        """Compare original and improved versions"""
        comparison = f"""
        IMPROVEMENT COMPARISON:

        CHANGES MADE:
        - Length: {len(original)} → {len(improved)} characters
        - Structure: Enhanced organization and flow
        - Clarity: Added explanatory examples
        - Completeness: Included additional relevant details
        - Engagement: Improved introduction and formatting

        KEY IMPROVEMENTS:
        1. Better opening that captures attention
        2. Clearer explanations of complex concepts
        3. More practical examples and applications
        4. Enhanced conclusion with actionable insights
        5. Overall better readability and structure

        CONFIDENCE LEVEL: High - Addresses all major critique points
        """

        return comparison

    def _format_quality_scores(self, metrics_obj: QualityMetrics) -> str:
        """Format quality scores for display"""
        output = []
        for criterion, score in metrics_obj.metrics.items():
            output.append(f"{criterion.capitalize()}: {score}/10")
        output.append(f"\\nOVERALL SCORE: {metrics_obj.overall_score():.1f}/10")
        return "\\n".join(output)

# Specialized improvers for different content types
class SummaryImprover(SelfReflectiveImprover):
    """Specialized improver for summaries"""

    def __init__(self):
        super().__init__()
        self.reflection_criteria.update({
            'conciseness': 'Is it appropriately concise without losing key information?',
            'key_points': 'Are the most important points clearly highlighted?'
        })

class CodeImprover(SelfReflectiveImprover):
    """Specialized improver for code explanations"""

    def __init__(self):
        super().__init__()
        self.reflection_criteria.update({
            'readability': 'Is the code easy to read and understand?',
            'efficiency': 'Is the solution efficient and well-optimized?',
            'best_practices': 'Does it follow coding best practices?'
        })

# Usage Examples and Demonstrations
def demonstrate_summary_improvement():
    """Demonstrate summary improvement with reflection"""
    improver = SummaryImprover()

    initial_summary = """
    Machine learning is a type of artificial intelligence that allows computers to learn
    and improve from experience without being explicitly programmed. It uses algorithms
    to find patterns in data and make predictions or decisions. There are different types
    like supervised learning, unsupervised learning, and reinforcement learning. It's used
    in many applications today.
    """

    print("=== SUMMARY IMPROVEMENT DEMO ===\\n")

    result = improver.improve_with_reflection(initial_summary, "summary")

    print("=== FINAL RESULTS ===")
    print(f"Final Quality Score: {result['final_quality_score']:.1f}/10")
    print(f"Improvement Iterations: {len(result['improvement_history'])}")

def demonstrate_code_explanation_improvement():
    """Demonstrate code explanation improvement"""
    improver = CodeImprover()

    initial_explanation = """
    This function sorts a list using bubble sort algorithm. It compares adjacent elements
    and swaps them if they are in the wrong order. This process is repeated until no
    more swaps are needed.
    """

    print("=== CODE EXPLANATION IMPROVEMENT DEMO ===\\n")

    result = improver.improve_with_reflection(initial_explanation, "code_explanation")

    print("=== FINAL RESULTS ===")
    print(f"Final Quality Score: {result['final_quality_score']:.1f}/10")

if __name__ == "__main__":
    print("Choose a demonstration:")
    print("1. Summary Improvement")
    print("2. Code Explanation Improvement")

    choice = input("Enter choice (1 or 2): ")

    if choice == "1":
        demonstrate_summary_improvement()
    elif choice == "2":
        demonstrate_code_explanation_improvement()
    else:
        print("Running both demonstrations...")
        demonstrate_summary_improvement()
        print("\\n" + "="*50 + "\\n")
        demonstrate_code_explanation_improvement()
```
