```python
import json
import time
from typing import Dict, List, Optional

class CustomerSupportChain:
    def __init__(self):
        self.conversation_state = {
            "stage": "initial_greeting",
            "customer_info": {},
            "issue_details": {},
            "resolution_attempts": []
        }

    def stage_1_greeting_and_triage(self, customer_input: str) -> Dict:
        """Initial greeting and issue identification"""
        prompt = f"""
        You are a friendly customer support representative.

        Customer message: "{customer_input}"

        Your tasks:
        1. Greet the customer warmly
        2. Identify the type of issue (technical, billing, account, product)
        3. Ask for basic information needed to help
        4. Set a helpful, professional tone

        Respond in JSON format:
        {{
            "response": "your response to customer",
            "issue_type": "technical/billing/account/product",
            "next_stage": "information_gathering",
            "required_info": ["list of info needed"]
        }}
        """
        return self._call_llm(prompt)

    def stage_2_information_gathering(self, customer_input: str, issue_type: str) -> Dict:
        """Gather specific information based on issue type"""
        prompt = f"""
        You are gathering information for a {issue_type} issue.

        Customer response: "{customer_input}"
        Previous issue type identified: {issue_type}

        Based on the issue type, ask specific clarifying questions:
        - Technical: Error messages, when it started, steps to reproduce
        - Billing: Account details, billing period, specific charges
        - Account: Account verification, specific account issues
        - Product: Product details, specific problems, usage context

        Respond in JSON format:
        {{
            "response": "your follow-up questions",
            "information_gathered": {{"key": "value"}},
            "next_stage": "solution_proposal",
            "confidence_level": "high/medium/low"
        }}
        """
        return self._call_llm(prompt)


    def _call_llm(self, prompt: str) -> Dict:
        """Simulate LLM call - replace with actual API call"""
        print(f"PROMPT SENT:\n{prompt}\n")

        if "greet the customer warmly" in prompt.lower():
            return {
                "response": "Hello! I'm here to help you today. I see you're having an issue. Could you please describe what's happening?",
                "issue_type": "technical",
                "next_stage": "information_gathering",
                "required_info": ["error_message", "when_started", "device_type"]
            }
        elif "gathering information for a" in prompt.lower():
            return {
                "response": "Thank you for that information. Could you please tell me what error message you're seeing and when this started happening?",
                "information_gathered": {"issue": "login_problem", "device": "mobile"},
                "next_stage": "solution_proposal",
                "confidence_level": "high"
            }

        return {"response": "Simulated response", "next_stage": "complete"}

# Usage Example
def run_customer_support_demo():
    support = CustomerSupportChain()

    print("=== CUSTOMER SUPPORT CHAIN DEMO ===\n")

    # Stage 1: Initial contact
    customer_message = "Hi, I can't log into my account on the mobile app"
    response1 = support.stage_1_greeting_and_triage(customer_message)
    print(f"STAGE 1 RESPONSE: {response1['response']}\n")

    # Stage 2: Information gathering
    customer_response = "I'm getting an 'Invalid credentials' error since yesterday"
    response2 = support.stage_2_information_gathering(customer_response, response1['issue_type'])
    print(f"STAGE 2 RESPONSE: {response2['response']}\n")




if __name__ == "__main__":
    run_customer_support_demo()
```
